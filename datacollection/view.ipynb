```python
import h5py
import matplotlib.pyplot as plt

def display_gameplay_sequence(filepath='pacman_rgb_gameplay_1_00_000.h5', start_index=0, num_samples=2):
    """
    Display sequences of (current_frame, action, next_frame) gameplay data.
    
    Args:
        filepath: Path to the HDF5 file
        start_index: Index from where to start displaying transitions
        num_samples: Number of frame-action-next_frame samples to display
    """
    try:
        with h5py.File(filepath, 'r') as f:
            frames = f['frames']
            next_frames = f['next_frames']
            
            actions = f['user_actions']
            ghost_actions = f['ghost_actions']
            rewards = f['rewards']
            resets = f['resets'] if 'resets' in f else None
            

            total_samples = len(frames)
            print(f"\nTotal samples available: {total_samples}")

            # Validate start_index
            if start_index >= total_samples:
                raise ValueError(f"Start index {start_index} is out of bounds.")

            # Adjust num_samples to stay within bounds
            num_samples = min(num_samples, total_samples - start_index)

            print(f"Displaying {num_samples} transitions starting from index {start_index}")

            # Create a figure with num_samples rows and 2 columns (current and next frames)
            fig, axes = plt.subplots(num_samples, 2, figsize=(12, 4 * num_samples))

            # Ensure axes is always a 2D array
            if num_samples == 1:
                axes = axes.reshape(1, 2)

            for i in range(num_samples):
                current_idx = start_index + i
                current_frame = frames[current_idx]
                next_frame = next_frames[current_idx]
                action = actions[current_idx]
                ghost_action = ghost_actions[current_idx]
                reward = rewards[current_idx]
                reset = resets[current_idx] if resets is not None else None

                # Plot current frame
             #   current_frame = current_frame[35:195, :160, :]
                axes[i, 0].imshow(current_frame)
                axes[i, 0].set_title(f'Current Frame {current_idx}\nAction: {action}, Ghost: {ghost_action} ,Reward: {reward:.2f}, Reset: {reset}')
                axes[i, 0].axis('off')

                # Plot next frame
             #   next_frame = next_frame[20:180, :160, :]
                axes[i, 1].imshow(next_frame)
                axes[i, 1].set_title(f'Next Frame {current_idx + 1}')
                axes[i, 1].axis('off')

            plt.tight_layout()
            plt.show()

            # Print dataset metadata
            print("\nDataset Metadata:")
            print(f"Collection date: {f.attrs['collection_date']}")
            print(f"Frame shape: {f.attrs['frame_shape']}")
            
            print("\nData Shapes:")
            print(f"Frames shape: {frames.shape}")
            print(f"Next Frames shape: {next_frames.shape}")
            print(f"Actions shape: {actions.shape}")
            print(f"Rewards shape: {rewards.shape}")
            if resets is not None:
                print(f"Resets shape: {resets.shape}")

    except FileNotFoundError:
        print(f"Error: Could not find file {filepath}")
    except KeyError as e:
        print(f"Error: Dataset not found in HDF5 file: {e}")
    except Exception as e:
        print(f"Error: {str(e)}")


if __name__ == "__main__":
    
    display_gameplay_sequence(filepath=r"", 
    start_index=3989, num_samples=10)
    # Red, pink, cyan, orange
```
